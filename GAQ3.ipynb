```python
import re

# Reglas iniciales (puedes modificarlas)
reglas = [
    "La contraseña debe tener mínimo 8 caracteres.",
    "Debe contener al menos una letra mayúscula.",
    "Debe contener al menos un número.",
    "Debe contener al menos un símbolo como *, #, $, %, &.",
    "No puede contener espacios.",
    "Debe contener al menos una palabra (ej: sol, blue, casa, cat...)."
]

def mostrar_reglas():
    print("\n📜 REGLAS ACTUALES:")
    for i, regla in enumerate(reglas, 1):
        print(f"{i}. {regla}")

def validar_contraseña(pwd):
    validaciones = [
        len(pwd) >= 8,                                     # Regla 1
        re.search(r"[A-Z]", pwd) is not None,             # Regla 2
        re.search(r"[0-9]", pwd) is not None,             # Regla 3
        re.search(r"[\*\#\$\%\&]", pwd) is not None,      # Regla 4
        " " not in pwd,                                   # Regla 5
        re.search(r"[a-zA-Z]{3,}", pwd) is not None       # Regla 6 (una palabra sencilla)
    ]
    return all(validaciones)

def agregar_reglas():
    print("\n🛠 Agregar una nueva regla:")
    nueva = input("Escribe la nueva regla: ")
    reglas.append(nueva)
    print("✅ Regla agregada exitosamente!")

def jugar():
    print("🎮 BIENVENIDO AL JUEGO DE LA CONTRASEÑA 🎮")
    mostrar_reglas()

    while True:
        print("\nElige una opción:")
        print("1. Intentar crear contraseña")
        print("2. Agregar una regla")
        print("3. Ver reglas")
        print("4. Salir")

        opcion = input("> ")

        if opcion == "1":
            pwd = input("\n🔐 Escribe la contraseña: ")
            if validar_contraseña(pwd):
                print("\n✅ CONTRASEÑA VÁLIDA! ¡Has ganado! 🎉")
            else:
                print("\n❌ Contraseña inválida. Revisa las reglas e inténtalo nuevamente.")

        elif opcion == "2":
            agregar_reglas()

        elif opcion == "3":
            mostrar_reglas()

        elif opcion == "4":
            print("\n👋 Gracias por jugar!")
            break

        else:
            print("\n⚠ Opción no válida, intenta otra vez.")

# Ejecutar el juego
jugar()
```
