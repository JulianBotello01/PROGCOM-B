```python
class Coreografia:

    # Paso 1 versión cadera
    def paso_1_cadera(self, a, b):
        for i in range(a):
            print("Muevo la cadera hacia la derecha")
        for i in range(a):
            print("Muevo la cadera hacia la izquierda")

    # Paso 1 versión brazos
    def paso_1_brazos(self, a):
        if a == 1:
            print("Levantar los brazos hacia arriba")
        elif a == 2:
            print("Levantar los brazos hacia arriba a la derecha")
        elif a == 3:
            print("Levantar los brazos hacia arriba a la izquierda")

    def paso_2(self, a):
        for i in range(a):
            print("Levanta las manos hacia arriba")

    def paso_3(self, a):
        for i in range(a):
            print("Pasa las manos por debajo de la quijada")

    def paso_4(self, a, b):
        for i in range(a):
            print("Se agacha")
        for i in range(b):
            print("Mueve los dos brazos levantándolos hacia los lados")

    def paso_5(self, a):
        for i in range(a):
            print("Realiza un giro")

    def paso_6(self, a):
        for i in range(a):
            print("Da un paso hacia atrás")

    def paso_7(self, a):
        for i in range(a):
            print("Flexiona las rodillas")

    def paso_8(self, a):
        for i in range(a):
            print("Extiende los brazos hacia adelante")

    def paso_9(self, a):
        for i in range(a):
            print("Extiende los brazos hacia los lados")

    def paso_10(self, a):
        for i in range(a):
            print("Lleva los brazos detrás de la cabeza")

    def paso_11(self, a):
        for i in range(a):
            print("Inclina el torso hacia adelante")

    def paso_12(self, a):
        for i in range(a):
            print("Inclina el torso hacia atrás")

    def paso_13(self, a):
        for i in range(a):
            print("Lleva brazos hacia abajo")

    def paso_14(self, a):
        for i in range(a):
            print("Lleva brazos hacia los hombros")

    def paso_15(self, direccion):
        if direccion == 1:
            print("Camina hacia la izquierda")
        elif direccion == 2:
            print("Camina hacia la derecha")
        else:
            print("Gira y vuelve al centro")

    # --- NUEVAS FUNCIONES PARA LA PARTE FINAL ---
    def seccion_1_movimientos_iniciales(self):
        print("SECCIÓN 1: Movimientos iniciales")
        print("Bailarina 1 levanta los brazos hacia la derecha")
        print("Bailarina 3 levanta los brazos hacia la izquierda")
        print("Bailarina 1 forma una 'L' con el brazo derecho")
        print("Bailarina 3 forma una 'L' con el brazo izquierdo")

    def seccion_2_mov_cuerpo_brazos(self):
        print("SECCIÓN 2: Movimiento del cuerpo y brazos")
        print("Bailarina 1 mueve el cuerpo y extiende el brazo a la izquierda")
        print("Bailarina 3 mueve el cuerpo y extiende el brazo a la derecha")
        print("Bailarina 1 levanta el codo a la derecha")
        print("Bailarina 3 levanta el codo a la izquierda")

    def seccion_3_coordinacion(self):
        print("SECCIÓN 3: Coordinación conjunta")
        print("Ambas extienden los brazos al centro formando una flecha hacia arriba")
        print("Dan un paso adelante y cruzan los brazos hacia la derecha")
        print("Rotan el antebrazo derecho")
        print("Giran el brazo izquierdo en sentido contrario a las agujas del reloj")

    def seccion_4_cambios_posicion(self):
        print("SECCIÓN 4: Cambios de posición")
        print("Bailarina 1 levanta los brazos")
        print("Bailarina 3 extiende los brazos hacia abajo")
        print("Intercambian las posiciones de los brazos")
        print("Extienden los brazos hacia los lados")
        print("Estiran los brazos arriba a la derecha")
        print("Mueven los brazos hacia la izquierda")

    def seccion_5_desplazamiento(self):
        print("SECCIÓN 5: Desplazamiento")
        print("Bailarina 1 da un paso a la izquierda y uno al frente")
        print("Bailarina 3 da un paso a la derecha y uno atrás")
        print("Ambas rotan el cuerpo hacia la derecha")

    def seccion_6_cierre(self):
        print("SECCIÓN 6: Cierre coreográfico")
        print("Bailarina 3 extiende los brazos a los lados")
        print("Bailarina 1 cruza los brazos")
        print("Proceden a hacer pasos indescriptibles")
        print(" Fin de la coreografía ")

    # --- FUNCIÓN QUE UNE TODAS LAS SECCIONES ---
    def paso_final(self):
        self.seccion_1_movimientos_iniciales()
        self.seccion_2_mov_cuerpo_brazos()
        self.seccion_3_coordinacion()
        self.seccion_4_cambios_posicion()
        self.seccion_5_desplazamiento()
        self.seccion_6_cierre()

    # --- EJECUCIÓN COMPLETA DEL BAILE ---
    def baile_completo(self):
        print(" INICIO DEL BAILE ")
        self.paso_1_cadera(2, 3)
        self.paso_1_brazos(1)
        self.paso_2(1)
        self.paso_3(1)
        self.paso_4(1, 1)
        self.paso_5(1)
        self.paso_6(1)
        self.paso_7(1)
        self.paso_8(1)
        self.paso_9(1)
        self.paso_10(1)
        self.paso_11(1)
        self.paso_12(1)
        self.paso_13(1)
        self.paso_14(1)
        self.paso_15(1)
        self.paso_final()
        print(" FIN DEL BAILE ")


# --- EJECUCIÓN ---
baile = Coreografia()
baile.baile_completo()
```
