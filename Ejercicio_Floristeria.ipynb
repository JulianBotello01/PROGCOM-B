```python
import random

def tienda():
    tienda_flores = {
        "frontal": {
            "Color": "amarillo",
            "puerta": 1,
            "ventanas": 0
        },
        "Lateral.I": {
            "Color": "amarillo",
            "puerta": 0,
            "ventanas": 1
        },
        "Lateral.D": {
            "Color": "amarillo",
            "puerta": 0,
            "ventanas": 1
        },
        "trasera": {
            "Color": "amarillo",
            "puerta": 0,
            "ventanas": 2
        },
        "superior": {
            "Color": "marron",
            "puerta": 0,
            "ventanas": 0
        }
    }
    return tienda_flores

def mostrar_parte_tienda():
    datos = tienda()

    print("Bienvenido a la floristeria 3 opciones")

    parte_solicitada = ""

    while parte_solicitada != "no quiero ver más":
        print("Partes disponibles de la tienda:")
        for parte in datos:
            print(f"- {parte}")
        print("- Escribe 'no quiero ver más' para salir")

        parte_solicitada = input("¿Qué parte de la tienda quieres ver?: ")

        if parte_solicitada == "no quiero ver más":
            print("Gracias por usar el programa.")
        elif parte_solicitada in datos:
            print("Características de la parte:", parte_solicitada)
            caracteristicas = datos[parte_solicitada]
            for clave, valor in caracteristicas.items():
                print(f"{clave}: {valor}")
            print()
        else:
            print("Esa parte no existe. Intenta otra vez.\n")


    entrar = input("¿Deseas entrar a la floristería para comprar flores? (si/no): ").lower()
    if entrar == "si":
        ejecutar_sistema_ventas()
    else:
        print("Hasta pronto 🌸")

#

def inventario():
    flores = ["tulipán", "lirio", "rosa", "dahlia", "margarita", "girasol", "clavel", "geranio"]
    seleccion = random.sample(flores, 3)

    inventario = {
        flor: {
            "precio": random.randint(2000, 6000),
            "cantidad": random.randint(10, 50),
            "vendidas": 0
        } for flor in seleccion
    }
    return inventario

def menu(a):
    print("\n---MENÚ DE FLORES DISPONIBLES FLORISTERÍA CARLITOS---")
    for flor, datos in a.items():
        print(f"{flor}: ${datos['precio']} cant: {datos['cantidad']} disponibles")
    print("------------------------------------------------------")

def ventas(inventario, flor, cantidad, total_ganado):
    if flor not in inventario:
        print(f"{flor} no se encuentra a la venta🤷‍♀️")
        return total_ganado
    datos = inventario[flor]
    if cantidad > datos["cantidad"]:
        print(f"Lo siento. No tenemos la cantidad requerida")
        return total_ganado

    datos["cantidad"] -= cantidad
    datos["vendidas"] = cantidad + datos["vendidas"]
    venta = datos["precio"] * cantidad
    total_ganado += venta
    print(f"Vendidas: {cantidad} de {flor} por un total de ${venta}")

    if datos["cantidad"] == 0:
        print(f"Acaba de comprar todas las unidades existentes de {flor} 🥀")
        del inventario[flor]
    return total_ganado

def cerrar_tienda(inventario_inicial, inventario_final, total_ganado):
    print("\n---Tienda Cerrada---")
    flores_vendidas = {}
    for flor, datos in inventario_inicial.items():
        cantidad_final = inventario_final.get(flor, {"cantidad": 0})["cantidad"]
        vendidas = datos["cantidad"] - cantidad_final
        flores_vendidas[flor] = vendidas

    mas_vendidas = max(flores_vendidas, key=flores_vendidas.get)
    menos_vendidas = min(flores_vendidas, key=flores_vendidas.get)

    print(f"Flor más vendida fue {mas_vendidas}: {flores_vendidas[mas_vendidas]} unidad(es)")
    print(f"Flor menos vendida fue {menos_vendidas}: {flores_vendidas[menos_vendidas]} unidad(es)")
    print(f"Total recaudado: ${total_ganado}")



def ejecutar_sistema_ventas():
    inventario_inicial = inventario()
    inventario_actual = {k: v.copy() for k, v in inventario_inicial.items()}
    total_ganado = 0

    comprando = "si"
    while comprando == "si" and inventario_actual:
        menu(inventario_actual)
        flor = input("¿Qué flor deseas comprar?: ").lower()
        try:
            cantidad = int(input("¿Cuántas unidades deseas?: "))
            total_ganado = ventas(inventario_actual, flor, cantidad, total_ganado)
        except ValueError:
            print("Cantidad inválida. Debe ser un número.")

        if inventario_actual:
            comprando = input("¿Deseas comprar otra flor? (sí/no): ").lower()

    cerrar_tienda(inventario_inicial, inventario_actual, total_ganado)



mostrar_parte_tienda()
```
