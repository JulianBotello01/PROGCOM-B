```python
from machine import Pin, I2C
import neopixel
import time
from mpu6050 import MPU6050
NUM_LEDS = 24
PIN_LEDS = 5
COLOR_BASE = (0, 0, 255)
np = neopixel.NeoPixel(Pin(PIN_LEDS), NUM_LEDS)
i2c = I2C(0, scl=Pin(22), sda=Pin(21))
mpu = MPU6050(i2c)
def actualizar_brillo():
    valores = mpu.get_values()
    intensidad = abs(valores['AcX']) + abs(valores['AcY']) + abs(valores['AcZ'])
    brillo = min(max(int(intensidad / 100), 10), 255)
    color = (
        int(COLOR_BASE[0] * brillo / 255),
        int(COLOR_BASE[1] * brillo / 255),
        int(COLOR_BASE[2] * brillo / 255)
    )
    for i in range(NUM_LEDS):
        np[i] = color
    np.write()
while True:
    actualizar_brillo()
    time.sleep(0.1)
```
