```python
import math

def colebrook(Re, D, e, tol=1e-6, max_iter=100):
    """
    Calcula el factor de fricción f usando la ecuación de Colebrook.
    """
    if Re < 2000:
        return 64/Re

    f = 0.02

    for _ in range(max_iter):
        rhs = -2.0 * math.log10((e/D)/3.7 + 2.51/(Re*math.sqrt(f)))
        f_new = 1.0 / (rhs**2)
        if abs(f_new - f) < tol:
            return f_new
        f = f_new

    raise ValueError("No hubo convergencia en la iteración de Colebrook.")

def perdida_carga(L, D, Q, rho, mu, e):
    """
    Calcula pérdidas de carga en una tubería.
    """
    A = math.pi * (D**2) / 4
    V = Q / A
    Re = rho * V * D / mu

    f = colebrook(Re, D, e)
    g = 9.81

    hf = f * (L/D) * (V**2) / (2*g)
    return hf, f, Re, V


print("=== CALCULADORA DE PÉRDIDAS DE CARGA EN TUBERÍAS ===")

try:
    L = float(input("Longitud de la tubería [m]: "))
    D = float(input("Diámetro de la tubería [m]: "))
    Q = float(input("Caudal volumétrico [m³/s]: "))
    rho = float(input("Densidad del fluido [kg/m³]: "))
    mu = float(input("Viscosidad dinámica del fluido [Pa·s]: "))
    e = float(input("Rugosidad absoluta de la tubería [m]: "))

    hf, f, Re, V = perdida_carga(L, D, Q, rho, mu, e)

    print("\n--- RESULTADOS ---")
    print(f"Número de Reynolds: {Re:.2f}")
    print(f"Velocidad del fluido: {V:.3f} m/s")
    print(f"Factor de fricción (f): {f:.5f}")
    print(f"Pérdida de carga (hf): {hf:.3f} m de columna de fluido")

except Exception as e:
    print("Error en los datos ingresados:", e)
```
